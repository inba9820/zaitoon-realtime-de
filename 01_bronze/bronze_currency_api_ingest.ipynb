```python
currency_df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("multiLine", "true")
    .option("cloudFiles.schemaLocation", "/Volumes/zaitoon_catalog/bronze/raw_events/_schemas/currency_api")
    .option("cloudFiles.inferColumnTypes", "true")
    .load("/Volumes/zaitoon_catalog/bronze/raw_events/api_currency")
)

(
    currency_df.writeStream
    .format("delta")
    .option("checkpointLocation", "/Volumes/zaitoon_catalog/bronze/raw_events/_checkpoints/currency_api")
    .trigger(availableNow=True)
    .toTable("zaitoon_catalog.bronze.currency_rates_api")
)
```
